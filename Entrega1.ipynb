```python
#Ejercicio 1
def consversortemperatura(valor):
  valornuevo = valor * (9/5) + 32
  return valornuevo
temperatura = float(input("Indroduzca el valor de la temperatur a convertir en farenheit:"))
temperatura_final = consversortemperatura(temperatura)
print(temperatura_final)
```

```python
#Ejercicio 2
def pagototal(consumo):
  propina = consumo * 0.15
  total = consumo + propina
  return total

consumo = float( input("Introducir total cuenta:"))
consumo_total = pagototal(consumo)
print(consumo_total)
```

```python
#Ejercicio 3
def verificadoredad(edad):
  if edad>=18:
    return "Mayor de edad"
  else:
    return "Menor de edad"

edad = int(input("Introducir edad a comprobar"))
comprobacion = verificadoredad(edad)
print(comprobacion)
```
